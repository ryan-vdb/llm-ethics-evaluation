### Import and Connect

```python
import duckdb
con = duckdb.connect("llm_ethics_data.duckdb")

def load_table(name):
    return con.sql(f"SELECT * FROM {name}").df()
```

##### Consistency Questions

```python
consistency_questions = load_table("consistency_questions")
display(consistency_questions)
```

##### Integrity Questions

```python
integrity_questions = load_table("integrity_questions")
display(integrity_questions)
```

##### Helpers

```python
helpers = load_table("helpers")
display(helpers)
```

##### Consistency Responses

```python
consistency_responses = load_table("consistency_responses")
display(consistency_responses)
```

##### Integrity Responses

```python
integrity_responses = load_table("integrity_responses")
display(integrity_responses)
```

##### Consistency Embeddings

```python
consistency_embeddings = load_table("consistency_embeddings")
display(consistency_embeddings)
```

##### Integrity Embeddings

```python
integrity_embeddings = load_table("integrity_embeddings")
display(integrity_embeddings)
```
